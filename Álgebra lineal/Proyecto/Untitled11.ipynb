```python
import numpy as np

# ---------- Matrices base ----------
A = np.array([[ 1, -1,  1],
              [ 0,  1, -1],
              [-1,  0,  1]])            # tablero: X = 1, O = -1, vacío = 0
u = np.ones(3, dtype=int)               # vector de unos
J = np.fliplr(np.eye(3, dtype=int))     # matriz antidiagonal

# ---------- Para mostrar X y O ----------
SIMBOLOS = {1: "X", -1: "O", 0: " "}

def mostrar_tablero(M):
    for i in range(3):
        print(" " + " | ".join(SIMBOLOS[v] for v in M[i]))
        if i < 2:
            print("---+---+---")

print("Tablero original:")
mostrar_tablero(A)
print("\nA =\n", A)

# Paso 1: suma de filas (A · u)
print("\nPaso 1 - Suma de filas (A·u):", A @ u)

# Paso 2: suma de columnas (uᵀ · A; NumPy trata u como fila)
print("Paso 2 - Suma de columnas (uᵀ·A):", u @ A)

# Paso 3: diagonal principal (traza de A)
print("Paso 3 - Diagonal principal, tr(A):", np.trace(A))

# Paso 4: diagonal secundaria (traza de A·J)
print("Paso 4 - Diagonal secundaria, tr(A·J):", np.trace(A @ J))
print("   A·J =\n", A @ J)

# Paso 5: ganador
sumas = np.concatenate([A @ u, u @ A, [np.trace(A), np.trace(A @ J)]])
if 3 in sumas:
    print("Paso 5 - Ganador: X")
elif -3 in sumas:
    print("Paso 5 - Ganador: O")
else:
    print("Paso 5 - No hay ganador")

# Paso 6: rotación 90° (Aᵀ · J)
R = A.T @ J
print("\nPaso 6 - Aᵀ =\n", A.T)
print("Paso 6 - R = Aᵀ·J =\n", R)

print("\nTablero rotado 90°:")
mostrar_tablero(R)
```
